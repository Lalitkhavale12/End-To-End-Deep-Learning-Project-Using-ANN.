```python
import pandas as pd 
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder,StandardScaler,LabelEncoder
import pickle
```

```python
data = pd.read_csv('Churn_Modelling.csv')
```

```python
data.head()
```

```python
data.columns
```

```python
data.drop(['RowNumber', 'CustomerId', 'Surname'],axis=1,inplace=True)
```

```python
data.head()
```

```python
data.info()
```

```python
data.describe()
```

```python
## Encoding the Gender 

label_encoder_gender = LabelEncoder()

data['Gender'] = label_encoder_gender.fit_transform(data['Gender'])
```

```python
data
```

```python
## Applying One hot encoding to the Geography


oh_encoder_geo = OneHotEncoder()
geo_encoded = oh_encoder_geo.fit_transform(data[['Geography']])
geo_encoded
```

```python
geo_encoded_df = pd.DataFrame(geo_encoded.toarray(),columns=oh_encoder_geo.get_feature_names_out(['Geography']))
```

```python
geo_encoded_df
```

```python
data = pd.concat([data.reset_index(drop=True),geo_encoded_df],axis=1)
```

```python
data.head()
```

```python
data.drop(['Geography'],axis=1,inplace=True)
```

```python
data.head()
```

```python
 # Independent and dependent features 

X = data.drop(['EstimatedSalary'],axis=1)
y = data['EstimatedSalary']
```

```python
X.head()
```

```python
y
```

```python
# train test and split

X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.20,random_state=42)
```

```python
scaler= StandardScaler()
```

```python
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
## Save the Encoders and scalers
with open('label_encoder_gender.pkl','wb') as file:
  pickle.dump(label_encoder_gender,file)

with open('onehot_encoder_geo.pkl','wb') as file:
  pickle.dump(oh_encoder_geo,file)

with open('scaler.pkl','wb') as file:
  pickle.dump(scaler,file)
```

### ANN model training for regression problem statement

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.callbacks import EarlyStopping,TensorBoard 
```

```python
# Build the model

model = Sequential([
  Dense(64,activation='relu',input_shape=(X_train_scaled.shape[1],)),
  Dense(32,activation='relu'),
  Dense(1) ## Output
])
```

```python
## Compile the model

model.compile(optimizer='adam',loss='mean_absolute_error',metrics=['mae'])
```

```python
model.summary()
```

```python
## Set up the TensorBoard

import datetime

log_dir = 'logs/fit/' + datetime.datetime.now().strftime("%Y%m%d-%H%M%S")

tensorflow_callback  = TensorBoard(log_dir=log_dir,histogram_freq=1)
```

```python
early_stopping_callback = EarlyStopping(monitor='val_loss',patience=10,restore_best_weights=True)
```

```python
model.fit(
  X_train_scaled,y_train,
  epochs=100,
  batch_size=32,
  validation_data=(X_test_scaled,y_test),
  callbacks=[tensorflow_callback,early_stopping_callback],
  verbose = 1
  )
```

```python
ann_acc = model.evaluate(X_test_scaled,y_test)[1]
```
